### Sum the number of interactions for a complex, per interaction

```python
interactions_count : dict[str, pd.DataFrame] = {}
for intr_name, intr_df in zip(interaction_names , interactions_dfs):
    intr_df = intr_df['activity_id'].value_counts().rename_axis('activity_id').reset_index(name=intr_name)
    # intr_df = intr_df.merge(mapping, on='activity_id', how='inner')
    intr_df = intr_df.merge(mapping, on='activity_id', how='right')
    intr_df[intr_name] = intr_df[intr_name].fillna(0).astype(int)
    # print(len(intr_df), len(mapping))
    assert len(intr_df) == len(mapping)
    interactions_count[intr_name] = intr_df

print(len(interactions_count))
interactions_count[interaction_names[0]].head()
```

```python
from kinodata.data import KinodataDocked
from kinodata.transform import TransformToComplexGraph

dataset = KinodataDocked(transform=TransformToComplexGraph(remove_heterogeneous_representation=False), use_multiprocessing=True, num_processes= 16)
dataset
```

Choose idents to check:

Examine each:
    <br> Keeping in mind that we can access graphs by index and not ident. So we need to retrieve the index from the ident list

```python
def check(dataset, bond_df, idents_list = [],
        num_samples = 5,
        ident_col_name = 'ident_processed',
        smiles_col_name = 'smiles_processed',
        activity_col_name = 'y_processed',
        ):
    """
    Checks if the info in merged dataframe by activity_id, matches the dataset:
        - checking ident
        - checking SMILES
    """
    # randome selection of n samples to check
    sample_idents = bond_df[ident_col_name].sample(num_samples, random_state=None).to_list()

    if not idents_list:
        print("Getting idents list")
        idents_list = dataset.data.ident.tolist() 
    for ident in sample_idents:
        # We can access the graphs in the dataset by index and not by ident, so:
        idx = idents_list.index(ident)
        print(f"Checking {ident= }, {idx= }")

        # Check SMILES
        assert dataset[idx].smiles == bond_df.loc[bond_df[ident_col_name] == ident, smiles_col_name].iloc[0], f"at {ident=} SMILES not matching"

        # Check activity value
        assert dataset[idx].y.item() == bond_df.loc[bond_df[ident_col_name] == ident, activity_col_name].iloc[0], f"at {ident=} activity value not matching"
```

# Relation between Interactions and Activity

```python
import matplotlib.pyplot as plt
```

```python
def compute_correlations(
    df: pd.DataFrame,
    bond_col: str,
    y_col: str = "y_processed",
) -> dict[str, float]:
    """Compute Pearson and Spearman correlations between number of bonds and affinity."""
    sub_df = df[[bond_col, y_col]].dropna()
    pearson = sub_df.corr(method="pearson").iloc[0, 1]
    spearman = sub_df.corr(method="spearman").iloc[0, 1]
    return {"pearson": pearson, "spearman": spearman}


def compute_bond_stats(
    df: pd.DataFrame,
    bond_col: str,
) -> dict[str, float]:
    """Compute basic statistics (mean, median, variance) for bond counts."""
    sub_series = df[bond_col].dropna()
    return {
        "mean": round(sub_series.mean(), 3),
        "median": round(sub_series.median(), 3),
        "variance": round(sub_series.var(), 3),
    }

try:
    from statsmodels.nonparametric.smoothers_lowess import lowess
    HAVE_LOWESS = True
except ImportError:
    HAVE_LOWESS = False
    print("statsmodels not installed: LOWESS trend will be skipped.")


def plot_scatter_with_trend(
    df: pd.DataFrame,
    interaction_name: str,
    bond_col: str,
    y_col: str = "y_processed",
):
    """Scatter plot + optional LOWESS trend."""
    sub_df = df[[bond_col, y_col]].dropna()
    x = sub_df[bond_col].to_numpy()
    y = sub_df[y_col].to_numpy()

    plt.figure(figsize=(6, 4))
    plt.scatter(x, y, alpha=0.4, s=10, label="data")

    if HAVE_LOWESS and len(sub_df) > 5:
        smooth = lowess(y, x, frac=0.3)
        plt.plot(smooth[:, 0], smooth[:, 1], linewidth=2, label="LOWESS trend", color="red")

    plt.xlabel(f"Number of {interaction_name}")
    plt.ylabel("Activity (y_processed)")
    plt.title(f"Activity vs {interaction_name}")
    plt.legend()
    plt.tight_layout()
    plt.show()


def plot_hexbin_density(
    df: pd.DataFrame,
    interaction_name: str,
    bond_col: str ,
    y_col: str = "y_processed",
):
    """Hexbin density plot for large datasets."""
    sub_df = df[[bond_col, y_col]].dropna()
    x = sub_df[bond_col].to_numpy()
    y = sub_df[y_col].to_numpy()

    plt.figure(figsize=(6, 4))
    hb = plt.hexbin(x, y, gridsize=40, mincnt=1)
    plt.colorbar(hb, label="count")
    plt.xlabel(f"Number of {interaction_name}")
    plt.ylabel("Activity (y_processed)")
    plt.title(f"Density of Activity vs {interaction_name}")
    plt.tight_layout()
    plt.show()
```

```python
for intr_name, intr_df in interactions_count.items():
    corrs = compute_correlations(intr_df, bond_col=intr_name)
    print(f"{intr_name} correlations: {corrs}")
    print(f"\nstats: {compute_bond_stats(intr_df, bond_col=intr_name)}")
    # plot_scatter_with_trend(intr_df, interaction_name=intr_name, bond_col=intr_name)\
    plot_hexbin_density(intr_df, interaction_name=intr_name, bond_col=intr_name)
```
